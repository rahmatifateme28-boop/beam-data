```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import random

# ============================================
# مشخصات فیزیکی تیر
# ============================================
L = 1.0
E = 200e9
I = 8.33e-9
rho = 7850
A = 0.001
EI = E * I
rhoA = rho * A

# ============================================
# پارامترهای عددی (اصلاح شده)
# ============================================
nx = 50
nt = 50000          # ✅ افزایش یافت
dx = L / nx
dt = 1e-6           # ✅ کاهش یافت

# ============================================
# بارگذاری و شرایط مرزی (اصلاح شده)
# ============================================
load_type = 'harmonic'
F0 = 1000.0         # ✅ افزایش یافت
omega_f = 140.0     # ✅ تنظیم شد
x_load = 0.5
BC = 'pinned-pinned'

# ============================================
# ترک‌های رندوم
# ============================================
num_cracks = random.randint(4, 5)
cracks = []
for _ in range(num_cracks):
    pos = random.uniform(0.1, 0.9)
    depth = random.uniform(0.1, 0.5)
    cracks.append((pos, depth))

print("تعداد ترک‌ها:", num_cracks)
print("موقعیت و عمق ترک‌ها:", cracks)

# ============================================
# تابع سختی موضعی
# ============================================
def get_EI_at_position(x_pos):
    EI_local = EI
    for crack_pos, crack_depth in cracks:
        if abs(x_pos - crack_pos * L) < dx/2:
            reduction = 1 - 0.5 * crack_depth
            EI_local *= reduction
    return EI_local

# ============================================
# حلقه اصلی
# ============================================
x = np.linspace(0, L, nx+1)
w = np.zeros((nx+1, nt+1))
w_prev = np.zeros(nx+1)
w_curr = np.zeros(nx+1)

for j in range(nt):
    f = np.zeros(nx+1)
    node_load = int(nx * x_load / L)
    
    if load_type == 'harmonic':
        f[node_load] = F0 * np.sin(omega_f * j * dt)
    else:  # impulse
        if j == 5:
            f[node_load] = F0 / dt
    
    w_new = np.zeros(nx+1)
    for i in range(2, nx-1):
        EI_local = get_EI_at_position(x[i])
        term1 = (EI_local * dt**2 / (rhoA * dx**4)) * (
            w_curr[i-2] - 4*w_curr[i-1] + 6*w_curr[i] - 4*w_curr[i+1] + w_curr[i+2]
        )
        term2 = 2*w_curr[i] - w_prev[i]
        term3 = (dt**2 / rhoA) * f[i]
        w_new[i] = term1 + term2 + term3
    
    w_new[0] = 0
    w_new[-1] = 0
    w_new[1] = -w_new[2]
    w_new[-2] = -w_new[-3]
    
    w_prev = w_curr.copy()
    w_curr = w_new.copy()
    w[:, j+1] = w_curr

# ============================================
# ذخیره‌سازی داده‌ها
# ============================================
points = [0.2, 0.35, 0.5, 0.65, 0.8]
data = {}
for p in points:
    idx = int(p * nx)
    data[f'x={p}'] = w[idx, :]

df = pd.DataFrame(data)
df.to_csv('beam_data.csv', index=False)
print("✅ داده‌ها در beam_data.csv ذخیره شدند.")

# ============================================
# رسم نمودارها
# ============================================
t = np.linspace(0, nt*dt, nt+1)

plt.figure(figsize=(12, 4))
plt.plot(t, w[int(0.5*nx), :], linewidth=2)
plt.xlabel('Time (sec)')
plt.ylabel('change loc (m)')
plt.title(f'answer-beam {load_type} - {num_cracks} ')
plt.grid()
plt.show()

plt.figure(figsize=(12, 4))
time_indices = [0, 5000, 10000, 15000, 20000, 25000]  # متناسب با nt جدید
for ti in time_indices:
    if ti < nt:
        plt.plot(x, w[:, ti], label=f't={ti*dt:.5f} s')
plt.xlabel('loc (m)')
plt.ylabel('change loc (m)')
plt.title('forms of beam')
plt.legend()
plt.grid()
plt.show()

print("✅ successful!")
```
